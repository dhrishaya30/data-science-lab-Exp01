In C and Java, variables have a fixed data type, so an int variable cannot be assigned a float value without an explicit conversion, causing a compile-time error. In Python, variables are dynamically typed, so the same variable can be reassigned to values of different data types at runtime without any compilation error.

```python
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

print(my_abs(5))    # 5
print(my_abs(-7))   # 7
print(my_abs(0))    # 0
```
