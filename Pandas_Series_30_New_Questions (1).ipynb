26. Analyse the following program. Determine the final result of total for every label and explain how Pandas index alignment determines each value.
north = Series({'Jan': 40, 'Feb': 55, 'Mar': 60})
south = Series({'Feb': 20, 'Mar': 25, 'Apr': 30})
total = north + south


```python
import pandas as pd

north = pd.Series({'Jan': 40, 'Feb': 55, 'Mar': 60})

south = pd.Series({'Feb': 20, 'Mar': 25, 'Apr': 30})

total = north + south

print(total)

# Jan → only in north → NaN
# Feb → 55 + 20 = 75
# Mar → 60 + 25 = 85
# Apr → only in south → NaN

# So, Pandas aligns the labels first and then performs the addition.
```

29. Write a short Pandas program that creates a labelled Series for four products and sales, displays its values and index, selects one label, filters sales above 500, checks whether a label exists, and converts the Series to a dictionary.

```python
import pandas as pd

sales = pd.Series([400, 650, 800, 300],
                  index=['Pen', 'Bag', 'Shoes', 'Book'])

print("Values:", sales.values)
print("Index:", sales.index)

print("Bag Sales:", sales['Bag'])

print("Sales above 500:")
print(sales[sales > 500])

print("Shoes exists:", 'Shoes' in sales.index)

sales_dict = sales.to_dict()
print("Dictionary:", sales_dict)
```
